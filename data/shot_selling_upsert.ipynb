```python
import pandas as pd
import pymysql
from sqlalchemy import create_engine, text
from datetime import datetime
import math

# =========================
# 0. DB 연결 설정
# =========================
DB_USER = ""
DB_PASSWORD ="" 
DB_HOST = ""
DB_PORT = 3306
DB_NAME = "qaima"

engine = create_engine(
    f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}?charset=utf8mb4"
)

# =========================
# 1. CSV 읽기
# =========================
csv_path = r"short_selling.csv"

# stock_code 앞자리 0 보존하려면 dtype=str 권장
df = pd.read_csv(
    csv_path,
    encoding="utf-8",
    dtype={"stock_code": str}
)

print("raw rows =", len(df))
display(df.head())
```

```python
print(df.columns.tolist())
```

```python
# =========================
# 3. stock_code 정규화 + 날짜 파싱 / DB안정화
# =========================
def normalize_stock_code(x):
    if pd.isna(x):
        return None
    s = str(x).strip().replace('"', "").replace("'", "")
    if s == "":
        return None
    if s.isdigit():
        s = s.zfill(6)
    return s.upper()

df["stock_code"] = df["stock_code"].apply(normalize_stock_code)
df["report_date"] = pd.to_datetime(df["report_date"], errors="coerce").dt.date

df["source"] = df["source"].fillna("KRX").replace("", "KRX")
df["source_screen_id"] = df["source_screen_id"].fillna("MDCSTAT301").replace("", "MDCSTAT301")

print("rows =", len(df))
display(df.head())
```

```python
dup_cnt = final_df.duplicated(subset=["stock_id", "report_date"]).sum()
print("duplicated(stock_id, report_date) =", dup_cnt)

if dup_cnt > 0:
    final_df = final_df.drop_duplicates(subset=["stock_id", "report_date"], keep="last").copy()
    print("after drop_duplicates rows =", len(final_df))
```

```python
insert_sql = """
INSERT INTO short_selling (
    stock_id,
    report_date,
    market_code,
    security_type,
    short_volume_total,
    short_volume_uptick_applied,
    short_volume_uptick_exempt,
    total_volume,
    short_volume_ratio,
    short_amount_total,
    short_amount_uptick_applied,
    short_amount_uptick_exempt,
    total_amount,
    short_amount_ratio,
    source,
    source_screen_id,
    created_at,
    updated_at
) VALUES (
    :stock_id,
    :report_date,
    :market_code,
    :security_type,
    :short_volume_total,
    :short_volume_uptick_applied,
    :short_volume_uptick_exempt,
    :total_volume,
    :short_volume_ratio,
    :short_amount_total,
    :short_amount_uptick_applied,
    :short_amount_uptick_exempt,
    :total_amount,
    :short_amount_ratio,
    :source,
    :source_screen_id,
    :created_at,
    :updated_at
)
"""

chunk_size = 5000
rows = final_df.to_dict(orient="records")

with engine.begin() as conn:
    for i in range(0, len(rows), chunk_size):
        chunk = rows[i:i+chunk_size]
        conn.execute(text(insert_sql), chunk)
        print(f"inserted {i + len(chunk)} / {len(rows)}")
```
